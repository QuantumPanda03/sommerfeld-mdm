```python
import numpy as np
from sommerfeld_mdm.sommerfeld import sommerfeld
from sommerfeld_mdm.mdm import mdm_quintupletQ0S0
from sommerfeld_mdm.abelian import yukawa


M = 10000
v_rel = 1e-6
Mw = 80.379
V_inf, V_eff, Gamma = mdm_quintupletQ0S0(M)
p = M * v_rel / 2
x_inf = 10.0 * p / Mw

T = sommerfeld(
    M=M,
    v_rel=v_rel,
    V_inf=V_inf,
    V_eff=V_eff,
    x_inf=x_inf
)

T_conj = T.conj().T

S3 = (
    (T_conj @ Gamma @ T)[2, 2].real
    / Gamma[2, 2]
)

print(f"Sommerfeld enhancement: S3 = {S3:.6f}")
```

```python
M = 1000.0
alpha = 0.01
mphi = 0.1
v_rel = 0.01

V_inf, V_eff = yukawa(alpha, mphi)

T = sommerfeld(
    M=M,
    v_rel=v_rel,
    V_inf=V_inf,
    V_eff=V_eff,
    x_inf=500
)

S = np.abs(T[0, 0])**2

# Analytic Coulomb limit
S_coulomb = (
    2*np.pi*alpha/v_rel
    / (1 - np.exp(-2*np.pi*alpha/v_rel))
)

print("Numerical Yukawa:", S)
print("Analytic Coulomb:", S_coulomb)
```
